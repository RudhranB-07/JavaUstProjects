```python
#Bar Graph for repsenting the approved vs rejected loan applications
approved=(data['Loan_Status']==1).sum()
rejected=(data['Loan_Status']==0).sum()
```

```python
labels=['Approved','Rejected']
values=[approved,rejected]
```

```python
plt.bar(labels,values,color=['#00000F','Teal','Dimgray'])
# plt.plot(values, color='black')
# plt.pie(values,labels=labels, colors=['#00000F','Teal'])
plt.xlabel('Loan Status')
plt.ylabel('Count of Applications')
plt.title('Loan Approval vs Rejection')
plt.show()
```

```python
Rural=(data['Property_Area']==0).sum()
Semiurban=(data['Property_Area']==1).sum()
Urban=(data['Property_Area']==2).sum()
labels=['Rural','Semiurban','Urban']
values=[Rural,Semiurban,Urban]
```

```python
plt.figure(figsize=(8,8))
plt.pie(values,colors=['#00000F','Teal','Dimgray'],startangle=90,autopct='%1.1f%%',textprops={'color':'white'}, wedgeprops={'edgecolor':'White', 'linewidth':2})
plt.legend(labels, loc='best')
plt.title('Property Area Classification')
plt.show()
```
