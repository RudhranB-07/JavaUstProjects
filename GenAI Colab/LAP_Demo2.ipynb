Building an ML model for Loan application classification



```python
#Importing Packages
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# reading csv
data = pd.read_csv("/content/LoanApprovalPrediction.csv")
```

```python
data.head()
```

```python
data.isnull().sum() # sum of null values against each column
```

```python
data['Dependents'] = data['Dependents'].fillna(data['Dependents'].mean()) #all missing vals (12 in this case) replcaed with coresspomding average
```

```python
data['LoanAmount']=data['LoanAmount'].fillna(data['LoanAmount'].mean())
```

```python
data['Loan_Amount_Term']=data['Loan_Amount_Term'].fillna(data['Loan_Amount_Term'].mean())
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data['Credit_History']=data['Credit_History'].fillna(data['Credit_History'].mode()[0])
```

```python
data.isnull().sum()
```

```python
data.head()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
data['Gender']=le.fit_transform(data['Gender'])
data['Married']=le.fit_transform(data['Married'])
data['Education']=le.fit_transform(data['Education'])
data['Self_Employed']=le.fit_transform(data['Self_Employed'])
data['Property_Area']=le.fit_transform(data['Property_Area'])
data['Loan_Status']=le.fit_transform(data['Loan_Status'])
```

```python
data = data.drop(['Loan_ID'],axis=1)
```

```python
data.head()
```

```python
#dividing the data into dependent and independent columnss
x=data.drop(['Loan_Status'],axis=1) #input independent column
y=data['Loan_Status'] #output dependent column
```

```python
#splitting the data into - training and testing
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.30)
#25% of the data will be used for testing and rest 75% data will be used for training purposes
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
model=LogisticRegression()
```

```python
model.fit(x_train,y_train) #training of the ML model on top of 75% of the data and by using Logistic Regression Algo takes place
```

```python
predictions=model.predict(x_test)
```

```python
#checking the accuracy of the model
from sklearn.metrics import accuracy_score
```

```python
accuracy_score=accuracy_score(y_test,predictions)
```

```python
print(accuracy_score)
```
